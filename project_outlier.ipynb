```python
# 등수가 1~8 밖에 있는 경우
df_total = df_total[df_total['Ranked'] != 0]
df_total['Ranked'].unique()
```

```python
# 중복행 개수가 8이 아닌 값 불러오기, 8이 아닌 개수 구하기
game_counts = df_total['gameId'].value_counts()
counts = game_counts[game_counts != 8]
# 결과를 화면에 출력합니다.
print(counts)
print(len(counts))
```

```python
# 1. 이상치(8명 미만 게임) 제거
df_total = df_total[df_total.groupby('gameId')['gameId'].transform('count') >= 8].copy()

# 2. 16개인 게임은 정확히 8개만 남기기
df_total = df_total.groupby('gameId').head(8).reset_index(drop=True)

# 결과 확인
print(len(df_total))
print(df_total['gameId'].nunique())
print(df_total['gameId'].value_counts().nunique() == 1) 
```

```python
# 게임 시간이 15분(900초) 미만 또는 60분(3600초) 초과인 비정상 게임 제거
df_total = df_total[(df_total['gameDuration'] >= 900) & (df_total['gameDuration'] <= 3600)]

print(len(df_total))
```

```python
# single quote(') → double quote(")로 바꿔서 json이 인식하게 만들기
import json
import re

def str_to_dict(s):
    if pd.isna(s):
        return {}
    s = str(s)
    s = re.sub(r"(\w+):", r'"\1":', s)    # 키에 따옴표 붙이기
    s = s.replace("'", '"')              # ' 를 " 로 바꾸기
    try:
        return json.loads(s)
    except:
        return {}

# 적용
df_total['champion_count'] = df_total['champion'].apply(str_to_dict)

# 딕셔너리의 '키 개수' (챔피언 종류 수)가 6개 이상인 행만 필터링
df_total = df_total[df_total['champion_count'].apply(len) >= 6].copy()

# 임시 컬럼 삭제
df_total.drop(columns='champion_count', inplace=True)

print(len(df_total))
```

```python
# 세트 3 시너지 제거
bad_traits = ['Avatar', 'Glacial', 'Wind', 'TemplateTrait', 'Desert', 'Mountain', 'Ocean', 
              'Summoner', 'Ranger', 'Metal', 'Druid', 'Alchemist', 'Inferno', 'Berserker', 
              'Soulbound', 'Crystal', 'Assassin', 'Warden', 'Woodland', 'Electric', 
              'Predator', 'Mage', 'Poison', 'Light', 'Shadow']

pattern = '|'.join(bad_traits)
df_total = df_total[~df_total['combination'].astype(str).str.contains(pattern, na=False)].copy()

print(len(df_total))
```
